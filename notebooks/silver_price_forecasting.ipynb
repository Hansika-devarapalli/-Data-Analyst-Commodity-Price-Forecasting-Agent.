```python
close_data = master[['Close']].copy()
close_data.head()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(close_data)

print(scaled_data.shape)
```

```python
import numpy as np

X = []
y = []

for i in range(60, len(scaled_data)-10):
    X.append(scaled_data[i-60:i, 0])
    y.append(scaled_data[i+10, 0])

X = np.array(X)
y = np.array(y)

print(X.shape)
print(y.shape)
```

```python
X = X.reshape(X.shape[0], X.shape[1], 1)

print(X.shape)
```

```python
split = int(0.8 * len(X))

X_train = X[:split]
X_test = X[split:]

y_train = y[:split]
y_test = y[split:]

print(X_train.shape)
print(X_test.shape)
```

```python
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python
model_lstm = Sequential([
    LSTM(50, return_sequences=False, input_shape=(60, 1)),
    Dense(25),
    Dense(1)
])
```

```python
model_lstm.compile(
    optimizer='adam',
    loss='mean_squared_error'
)
```

```python
history = model_lstm.fit(
    X_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_data=(X_test, y_test)
)
```

```python
lstm_predictions = model_lstm.predict(X_test)
```

```python
lstm_predictions = scaler.inverse_transform(lstm_predictions)

y_test_actual = scaler.inverse_transform(
    y_test.reshape(-1, 1)
)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np

mae = mean_absolute_error(y_test_actual,lstm_predictions)
rmse = np.sqrt(mean_squared_error(y_test_actual, lstm_predictions))
r2 = r2_score(y_test_actual,lstm_predictions)

print("MAE:", mae)
print("RMSE:", rmse)
print("R2:", r2)
```

```python
comparison_lstm = pd.DataFrame({
    "Actual": y_test_actual.flatten(),
    "Predicted": lstm_predictions.flatten()
})

comparison_lstm.head(10)
```

```python
import numpy as np

final_dashboard["Signal"] = np.where(
    final_dashboard["LSTM_Prediction"] > final_dashboard["Actual_Price"],
    "BUY",
    "SELL"
)
```

```python
capital = 10000
cash = capital
silver_holding = 0

portfolio_value = []

for i in range(len(final_dashboard)):

    signal = final_dashboard["Signal"].iloc[i]
    price = final_dashboard["Actual_Price"].iloc[i]

    if signal == "BUY" and cash >= price:
        silver_holding += 1
        cash -= price

    elif signal == "SELL" and silver_holding > 0:
        silver_holding -= 1
        cash += price

    total_value = cash + silver_holding * price
    portfolio_value.append(total_value)

final_dashboard["Portfolio_Value"] = portfolio_value
```
